```python
numbers = [1,2,3,4,5,6]
even_num = [num ** 2 for num in numbers if num % 2 == 0] #list comprehension#
print(even_num)
```

```python
student = {
    "name" : "Rikesh",
    "age" : 20
}
student.update({
    "grade" : "A",
    "age" : 21
})

print (student)
student.clear()
print(student)
student.setdefault("age","22")
print(student["age"])
```

```python
students = {
    "student1": {"name": "Rikesh", "age": 20},
    "student2": {"name": "Ram", "age": 21}
}

students["student1"]["age"] = 22
students["student2"]["grade"] = "A"

for student_id, details in students.items():
    print(details["name"], details["age"])
```

```python
numbers = {num : num ** 2
           for num in range(1,11) 
           if num % 2 == 0
          }
print(numbers)
```

```python
numbers = {
    num: num ** 2
    for num in range(1, 11)
    if num % 2 == 0
}

print(numbers)
```

```python
numbers = {x: x**2 for x in range(1, 6) if x % 2 != 0}

numbers[2] = 4
del numbers[3]

print(numbers)
print(len(numbers))
numbers.clear()
print(numbers)
```
